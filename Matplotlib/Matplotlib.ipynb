```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x = np.linspace(0,5,11)
x
```

```python
y = x ** 2
```

```python
y
```

```python
plt.plot(x,y)
plt.xlabel('X Label')
plt.ylabel('Y Label')
plt.title('Title')
plt.show()
```

```python
fig = plt.figure(figsize=(10,3),dpi=100)
axis1 =  fig.add_axes([0.1,0.1,0.7,0.8])
axis1.plot(x,y)
axis2 = fig.add_axes([0.1,0.1,0.5,0.5]) #left,right,height,width (0-1)
axis2.plot(y,x)
```

```python
df = sns.load_dataset('tips')
df
```

```python
df['size'].unique()
```

```python
sns.distplot(df['total_bill'],kde=True) #histplot
```

```python
plt.subplot(1,2,1)
sns.distplot(df['total_bill'],kde=True) #histplot
plt.subplot(1,2,2)
sns.distplot(df['total_bill'],kde=False)
```

```python
sns.jointplot(x='total_bill',y='tip',data=df,kind='scatter')
```

```python
sns.pairplot(df,hue='sex')
```

```python
sns.countplot(x=df['sex'],hue = df['smoker'])
```

```python
import numpy as np
sns.barplot(x=df['sex'],y=df['tip'],estimator=np.median)
```

```python
import numpy as np
sns.barplot(x=df['sex'],y=df['total_bill'],estimator=np.median)
```

```python
sns.boxplot(x=df['day'],y=df['total_bill'])
```

```python
sns.lmplot(x='total_bill',y='tip',data=df,hue='sex',col='day',row='time')
```
