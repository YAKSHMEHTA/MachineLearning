```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11)
x
```

```python
y = x ** 2
```

```python
y
```

```python
plt.plot(x,y)
plt.xlabel('X Label')
plt.ylabel('Y Label')
plt.title('Title')
plt.show()
```

```python
fig = plt.figure(figsize=(10,3),dpi=100)
axis1 =  fig.add_axes([0.1,0.1,0.7,0.8])
axis1.plot(x,y)
axis2 = fig.add_axes([0.1,0.1,0.5,0.5]) #left,right,height,width (0-1)
axis2.plot(y,x)
```

```python
plt.scatter(x,y)
```
